```python
import sqlite3

conn = sqlite3.connect("data.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks REAL
)
""")

conn.commit()

conn.close()

print("Table created")
```

```python
import sqlite3

conn = sqlite3.connect("data.db")

cursor = conn.cursor()

cursor.execute(
    "INSERT INTO students (name, age, marks) VALUES (?, ?, ?)",
    ("Aditya", 21, 85.5)
)

conn.commit()

conn.close()

print("Data inserted")
```

```python
import sqlite3

conn = sqlite3.connect("data.db")

cursor = conn.cursor()

cursor.execute("SELECT * FROM students")

rows = cursor.fetchall()

for row in rows:
    print(row)

conn.close()
```

```python
import sqlite3

conn = sqlite3.connect("data.db")
cursor = conn.cursor()

cursor.execute(
    "SELECT * FROM students WHERE marks > ?",
    (80,)
)

rows = cursor.fetchall()

for row in rows:
    print(row)

conn.close()
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("data.db")

df = pd.read_sql(
    "SELECT * FROM students",
    conn
)

print(df)

conn.close()
```
